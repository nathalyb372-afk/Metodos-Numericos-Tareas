### 1. Use el método de bisección para encontrar soluciones precisas dentro de $10^{-2}$ para $x^{3} - 7x^{2} + 14x - 6 = 0 $ en cada intervalo

Definir la función en python para el método de la bisección

```python
def bisectriz(funcion, limite_inferior, limite_superior, tolerancia_error):
    if funcion(limite_inferior) * funcion(limite_superior) >= 0:
        raise ValueError("f(a) y f(b) deben tener signos opuestos.")
    while (limite_superior - limite_inferior) / 2.0 > tolerancia_error:
        punto_medio = (limite_inferior + limite_superior) / 2.0
        if funcion(punto_medio) == 0:
            print( "La raíz exacta es:", punto_medio)  
        elif funcion(limite_inferior) * funcion(punto_medio) < 0:
            limite_superior = punto_medio
        else:
            limite_inferior = punto_medio
    print( "La raíz aproximada es:", (limite_inferior + limite_superior) / 2.0)
    return (limite_inferior + limite_superior) / 2.0
```

Definimo la función

```python
def funcion_polinomial(variable_x):
    return variable_x**3 - 7*variable_x**2 + 14*variable_x - 6
```

##### a) $$[0 , 1]$$

```python
bisectriz(funcion_polinomial, 0, 1, 10**-2)
```

##### b) $$[1 , 3.2]$$

```python
bisectriz(funcion_polinomial, 1, 3.2, 10**-2)
```

##### c) $$[3.2 , 4]$$

```python
bisectriz(funcion_polinomial, 3.2, 4,10**-2)
```

#### 1. Un abrevadero de longitud $L$ tiene una sección transversal en forma de semicírculo con radio r. Cuando se llena con agua hasta una distancia h a partir de la parte superior, el volumen $V$ de agua es $$V = L [0.5\pi r^2 - r^2arcsen(\frac{h}{r}) - h(r^2 - h^2)^{\frac{1}{2}}]$$

Suponga que: $L = 10cm, r = 1 cm, V = 12.4 cm^3$. Encuentre la profundidad del agua en el abrevadero dentro de 0.01 cm

Sabemos que la profundidad va a estar dado por:

$$Profundidad = r - h$$

Se reescribe la ecuación en términos de h $$12.4 = 10[0.5\pi 1^2 - 1^2 arcsen(\frac{h}{1}) - h(1^2 - h^2)^{\frac{1}{2}}]$$


$$y = 10[0.5\pi - arcsen(h) - h(1 - h^2)^{\frac{1}{2}}] - 12.4$$

Definimos la función en python

```python
def funcion_trigonometrica_compleja(variable_x):
    return 10*(0.5*math.pi - np.arcsin(variable_x) - variable_x * np.sqrt(1 - variable_x**2)) - 12.4
```

Encontramos la raíz en el intervalo $[0,1]$

```python
bisectriz(funcion_trigonometrica_compleja, 0, 1, 0.01)
```

Calculamos la profundidad

```python
valor_profundidad = 1 - float(bisectriz(funcion_trigonometrica_compleja, 0, 1, 0.01))
print("La profundidad es:", valor_profundidad)
```

#### 2. Un objeto que cae verticalmente a través del aire está sujeto a una resistencia viscosa, así como a la fuerza de gravedad. Suponga que un objeto con masa $m$ cae desde una altura $s_0$ y que la altura del objeto después de t segundos es $$s(t) = s_0 - \frac{mg}{k}t + \frac{m^2g}{k^2}(1-e^{\frac{-kt}{m}})$$ donde $g = 9.81 m/s^2$ y $k$ representa el coficiente de resistnecia del aire en $Ms/m$. Suponga $s_0 = 300m$, $m = 0.25 kg$ y $k = 0.1 Ns/m$. Encuentre, dentro de 0.01 segundos, el tiempo que tarda un cuarto de kg en golpear el piso.

Reescribimos la ecuación con los valores conocidos

 $$s(t) = 300 - \frac{0.25(9.81)}{0.1}t + \frac{0.25^2(9.81)}{0.1^2}(1-e^{\frac{-0.1t}{0.25}})$$

 $$s(t) = 300 - 25.525t + 61.3125(1-e^{\frac{-0.1t}{0.25}})$$

Definimos la función

```python
def funcion_modelo_exponencial(variable_x):
    return 300 - 25.525*variable_x + 61.3125*(1 - math.exp(-0.1*variable_x/0.25))
```

Encontramos las raíces

```python
print('El tiempo que tarda en golpear el piso es:',
       bisectriz(funcion_modelo_exponencial, 0, 20, 0.01), 'segundos')
```
